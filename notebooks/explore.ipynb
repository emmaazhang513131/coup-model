```python
def all_states():
    return list(itertools.product(
        range(0, COIN_CAP+1),   # c1
        range(0, COIN_CAP+1),   # c2
        range(1, MAX_LIVES+1),   # l1
        range(1, MAX_LIVES+1),   # l2
        [PLAYER_1, PLAYER_2]
    ))
```

```python
def legal_actions(c1, c2, turn):
    actions = []
    if turn == PLAYER_1:
        if c1 >= 10:
            actions.append("coup")
            return actions
        actions.append("income")
        if c2>0:
            actions.append("steal")
        if c1 >= 7 and c1<10:
            actions.append("coup")
            return actions
    else:
        if c2 >= 10:
            actions.append("coup")
            return actions
        actions.append("income")
        if c1>0:
            actions.append("steal")
        if c2 >= 7 and c2<10:
                actions.append("coup")
                return actions
    return actions
    
```
